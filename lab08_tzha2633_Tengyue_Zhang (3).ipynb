#Function to delete stop words

```python
import nltk                          
nltk.download('stopwords')

from nltk.corpus import stopwords
stopwords = stopwords.words('english')
```

```python
#View the data type of stopwords
type(stopwords)
```

```python
stopwords[:2]
```

```python
#Remove the part of non-English letters and replace with |
import re
def remove_stop_words(line, stopwords):
    compiler = re.compile("[^a-zA-Z ]")
    repl = compiler.sub(" | ", line)
    repl = repl.lower()
    s = ""
    for word in repl.split(" "):
        if word in stopwords:
            s = s + "|"
        else:
            s = s + " " + word
        
    return s.strip("|")
```

#Function to remove page numbers with spaces

```python
#Replace page numbers with spaces
def flat_clause(line):
    compiler = re.compile("Page \d+[-]*\d*")
    repl = compiler.sub("", line)
    return repl.split("()")
```

# Candidate Phrase Identification


```python
#To get candidate phrases
def get_candidate_words(line, max_length):
    lst = line.split("|")
    res_lst = []
    for word in lst:
        word = word.strip()
        if word != "":
            word_lst = word.split(" ")
            word_lst = [item for item in word_lst if item != ""]
            if len(word_lst) <= max_length:
                res_lst.append(" ".join(word_lst))
    return res_lst
```

#Function to compute the score of word, the score of phrase, and return candidate phrases of each document that required



```python
# Computing the score of word
def compute_word_score(candidate_lst):
    split_candidate_lst = []
    for item in candidate_lst:
        split_candidate_lst.extend(item.split(" "))
    candidate_words = list(set(split_candidate_lst)) 
    fre_dic = {}
    deg_dic = {}
    for word in candidate_words:
        # Computing the frequency of word
        fre_dic[word] = split_candidate_lst.count(word)
        # Computing the degree of word
        degree = 0
        for items in candidate_lst:
            if word in items:
                degree += len(items.split(" "))
        deg_dic[word] = degree      
    score_dic = {}
    for word in candidate_words:
        score_dic[word] = deg_dic[word]/fre_dic[word]
    return [candidate_lst, score_dic]


# Computing the score of phrase
def compute_phrase_score(candidate_lst, word_score):
    phrase_score_dic = {}
    for phrase in candidate_lst:
        phrase_score_dic[phrase] = 0
        for word in phrase.split(" "):
            phrase_score_dic[phrase] += word_score[word]
    return phrase_score_dic

# Return candidate phrases of each document that required
def top_n_candidate(score_dic, n):
    res = sorted(score_dic.items(), key=lambda x: x[1], reverse=True)
    top_n = []
    for i in range(min(len(score_dic), n)):
        top_n.append(res[i][0])
    return top_n    
```

```python
def candidate_list_word(candidate_lst):
    res = []
    for item in candidate_lst:
        for word in item.split(" "):
            res.append((item, word))
    return res

def compute_degree(phrase):
    words = phrase.split(" ")
    words = [item for item in words if item != ""]
    return len(words)
```

```python
#Split data into combinations of (phrases, words)
candidate_list_word_rdd = clauses_candidate_word_rdd.flatMap(lambda x:candidate_list_word(x))
```

```python
candidate_list_word_rdd.take(5)
```

```python
#To calculate the frequency of each word
fre_rdd = candidate_list_word_rdd.map(lambda x:(x[1], 1)).reduceByKey(lambda x,y:x+y)
```

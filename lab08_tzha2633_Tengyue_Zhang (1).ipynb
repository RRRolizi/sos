#Calculate freq(w)
The number of times it occurs in the document. 

```python
from collections import defaultdict
def getDeg(phrases):
    wordDegree = defaultdict(int)
    for phrase in phrases:
        pLength = len(phrase)
        for i in range(pLength):
            wordDegree[phrase[i]] += pLength 
    return wordDegree
```
